```python
import difflib
from langchain_ollama import OllamaLLM
import ast
import re

model = OllamaLLM(model="llama3.2")


def find_common_substrings(str1, str2):
    common = set()
    len1, len2 = len(str1), len(str2)
    newCommon = set()

    # Generate all substrings of str1
    for i in range(len1):
        for j in range(i + 1, len1 + 1):
            substring = str1[i:j]
            if substring in str2:
                common.add(substring)
    for i in common:
        if len(i) >= 4:
            newCommon.add(i)
    return newCommon

def longest_common_substring(str1, str2):
    seq_match = difflib.SequenceMatcher(None, str1, str2)
    match = seq_match.find_longest_match(0, len(str1), 0, len(str2))

    if match.size > 0:
        return str1[match.a: match.a + match.size]
    return ""

def askOllama(prompt):
    result = model.invoke(input=prompt)
    return result

def cleanOllamaOutput(output):
    pattern = r"\[.*?\]"
    
    matches = re.findall(pattern, output, re.DOTALL)
    a = matches[0].replace("\n", "")
    escaped_string = re.sub(r"(?<=\w)'(?=\w)", r"\'", a)
    return ast.literal_eval(escaped_string.replace("\\n", "").replace("\\\\", "\\"))

chat = ['a challenging patient', 'history of uncooperative behavior', 'noncompliant, citing his preferences for junk food', 'frequent cursing at nurses during previous medical consultations', 'Noncompliant with nutritional recommendations', 'Frequent episodes of cursing at healthcare staff', 'excessive time around local fast food establishments', 'notable for shorter stature than average for age', 'a difficult patient', 'challenging interactions and unreliable historical recounting', 'refused several exams', 'resistant to modifying his diet', 'cursing at staff persisted', 'his noncompliance has been a significant barrier to his care']
answer = ['challenging patient', 'uncooperative behavior', 'noncompliant', 'citing his preferences for junk food', 'frequent cursing at nurses', 'difficult patient', 'challenging interactions', 'unreliable historical recounting', 'refused several exams', 'resistant to modifying his diet', 'complicating our attempts to provide care', 'significant barrier to his care', 'noncompliance has been a significant barrier to his care', "Mohammad's noncompliance has been a significant barrier to his care"]

matches = []

while True:
    try:
        print(cleanOllamaOutput(askOllama("Return to me only a list of elements which are referring to the same thing in these two lists in JSON format. Here are the two lists: " + str(chat) + ", " + str(answer))))
        break
    except:
        print()


```
